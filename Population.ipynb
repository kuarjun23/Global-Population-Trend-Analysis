### Cleaning the 'Yearly Change %' Column

```python
df["Yearly Change %"] = df["Yearly Change %"].astype(str).str.replace("%", "").str.replace("âˆ’", "-").str.strip()
df["Yearly Change %"] = pd.to_numeric(df["Yearly Change %"])
```

## Cleaning the 'Yearly Change' Column

The 'Yearly Change' column, which is currently an object type, needs to be converted to a numeric data type for proper analysis. This involves removing any non-numeric characters like commas and non-standard minus signs.

```python
df['Yearly Change'] = df['Yearly Change'].astype(str).str.replace(',', '').str.replace('âˆ’', '-').str.strip()
df['Yearly Change'] = pd.to_numeric(df['Yearly Change'])
```

## Cleaning the 'Migrants_Net' Column

Similar to the previous cleaning steps, the 'Migrants_Net' column is currently an object type due to special characters. This column needs to be converted to a numeric data type by removing commas and replacing non-standard minus signs.

```python
df['Migrants_Net'] = df['Migrants_Net'].astype(str).str.replace(',', '').str.replace('âˆ’', '-').str.strip()
df['Migrants_Net'] = pd.to_numeric(df['Migrants_Net'])
```

## Cleaning the 'Urban Population %' Column

The 'Urban Population %' column is an object type and contains percentage symbols. To use this column for numerical analysis, the percentage symbol needs to be removed, and the column needs to be converted to a numeric (float) data type.

```python
df['Urban Population %'] = df['Urban Population %'].astype(str).str.replace('%', '').str.replace('âˆ’', '-').str.replace('â€“', '-').str.strip()
df['Urban Population %'] = pd.to_numeric(df['Urban Population %'], errors='coerce')
```

## Cleaning the 'Urban Population' Column

The 'Urban Population' column is currently stored as an object type, likely due to the presence of commas as thousand separators. To perform numerical analysis, this column needs to be converted to a numeric data type by removing these commas.

```python
df['Urban Population'] = df['Urban Population'].astype(str).str.replace(',', '').str.replace('â€“', '-').str.strip()
df['Urban Population'] = pd.to_numeric(df['Urban Population'], errors='coerce')
```

## Rename the World Share column to World Share %

```python
# Renaming the column for clarity
df.rename(columns={'World Share': 'World Share %'}, inplace=True)
print(" 'World Share' column renamed to 'World Share %'. ")
```

## Correcting 'World Share %' to Decimal (0-1) Representation

The 'World Share %' column is currently on a percentage scale (e.g., 17.78 instead of 0.1778). I will now divide this column by 100 to convert it to a decimal representation (0-1).

```python
# Converting 'World Share %' to decimal format by dividing by 100
df['World Share %'] = df['World Share %'] / 100
print("'World Share %' column successfully converted to decimal format (0-1).")
```

## Rechecking Data Types After Cleaning

Now that several columns have been cleaned and converted to numeric types, we should recheck the data types of all columns to ensure that the cleaning process was successful and that all numerical columns are correctly represented.

```python
df.info()
```

## Handling Missing Values in 'Urban Population %' and 'Urban Population'

After converting 'Urban Population %' and 'Urban Population' columns to numeric types using `errors='coerce'`, `NaN` values were introduced where conversion failed. These missing values need to be addressed before further analysis.

A common strategy for numerical data is to fill missing values with a measure of central tendency, such as the median. The median is preferred over the mean as it is less sensitive to outliers. I will fill these `NaN` values with the median of their respective columns.

```python
print("Missing values before imputation:")
df[['Urban Population %', 'Urban Population']].isnull().sum()
```

```python
# Re-apply cleaning for 'Urban Population %' to ensure numeric type with NaNs
df['Urban Population %'] = df['Urban Population %'].astype(str).str.replace('%', '').str.replace('âˆ’', '-').str.replace('â€“', '-').str.strip()
df['Urban Population %'] = pd.to_numeric(df['Urban Population %'], errors='coerce')

# Re-apply cleaning for 'Urban Population' to ensure numeric type with NaNs
df['Urban Population'] = df['Urban Population'].astype(str).str.replace(',', '').str.replace('â€“', '-').str.strip()
df['Urban Population'] = pd.to_numeric(df['Urban Population'], errors='coerce')

# Calculate the median for 'Urban Population %' and 'Urban Population'
median_urban_pop_perc = df['Urban Population %'].median()
median_urban_pop = df['Urban Population'].median()

# Fill NaN values with the calculated medians
df['Urban Population %'] = df['Urban Population %'].fillna(median_urban_pop_perc)
df['Urban Population'] = df['Urban Population'].fillna(median_urban_pop)

print(f"'Urban Population %' NaN values filled with median: {median_urban_pop_perc:.2f}")
print(f"'Urban Population' NaN values filled with median: {median_urban_pop:.2f}")
```

## Rechecking Missing Values After Imputation

It's important to verify that all `NaN` values have been successfully handled in the 'Urban Population %' and 'Urban Population' columns.

```python
print("Missing values after imputation:")
df[['Urban Population %', 'Urban Population']].isnull().sum()

# Also rechecking overall info to confirm all columns are clean
df.info()
```

## Creating Derived Columns

Derived columns, also known as feature engineering, involve creating new features from existing ones. This process can help uncover deeper insights, improve model performance, and simplify analysis by capturing relationships that aren't immediately apparent in the raw data.

I will create the following derived columns:

1.  **Net Migration Rate**: This indicates the impact of migration on the total population, relative to the population size.
2.  **Rural Population**: The difference between the total population and the urban population.
3.  **Rural Population %**: The percentage of the total population living in rural areas.
4.  **Years Since 1955**: A simple time-based feature to represent the progression of years from the start of our dataset, useful for time-series analysis.

```python
# Ensure 'Migrants_Net' and 'Yearly Change' are numeric types
df['Migrants_Net'] = df['Migrants_Net'].astype(str).str.replace(',', '').str.replace('âˆ’', '-').str.strip()
df['Migrants_Net'] = pd.to_numeric(df['Migrants_Net'])

df['Yearly Change'] = df['Yearly Change'].astype(str).str.replace(',', '').str.replace('âˆ’', '-').str.strip()
df['Yearly Change'] = pd.to_numeric(df['Yearly Change'])

# 1. Net Migration Rate
df['Net Migration Rate'] = (df['Migrants_Net'] / df['Population'])

# 2. Rural Population
df['Rural Population'] = df['Population'] - df['Urban Population']

# 3. Rural Population %
df['Rural Population %'] = (df['Rural Population'] / df['Population'])

# 4. Years Since 1955
df['Years Since 1955'] = df['Year'] - 1955

print("New derived columns created successfully!")
```
